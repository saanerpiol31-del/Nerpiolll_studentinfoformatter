```python
# Ask user basic information - Ver 2
first_name = input("Enter first name: ")
last_name = input("Enter last name: ")
section = input("Enter section: ")
favorite_subject = input("Enter favorite subject: ")

#Combine the first and last name
full_name = first_name + " " + last_name

#Convert full name to upper case
name_code = full_name.upper()

#Convert the section to upper case
section2 = section.upper()

#To capitalize the first letter  of every word
favorite_Subject2 = favorite_subject.title()

#To display result
print("\nformatted output: ")
print("Name:", full_name)
print("Name code:", name_code)
print("Section:", section2)
print("Favorite Subject:", favorite_Subject2)

#Count the characters in the full name
print("Character count:", len(full_name))
```
